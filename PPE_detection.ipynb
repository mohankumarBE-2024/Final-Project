```python
from ultralytics import YOLO

model = YOLO('yolov8n.pt')

results = model.train(
    data='/content/dataset/data.yaml',
    epochs=30,          
    imgsz=640,          
    batch=16,           
    device=0,           
    project='Intelliguard_PPE',
    name='yolov8_nano_run1'
)
```

```python
import os

print("Searching for your trained weights...")
found = False

for root, dirs, files in os.walk('/content'):
    for file in files:
        if file == 'best.pt':
            print(f"\nFound it! The exact path is:\n'{os.path.join(root, file)}'")
            found = True

if not found:
    print("\nCould not find 'best.pt'. Did the training run finish completely without errors?")
```

```python
import shutil

source_path = '/content/runs/detect/Intelliguard_PPE/yolov8_nano_run1/weights/best.pt'

destination_path = '/content/drive/MyDrive/intelliguard_best_v1.pt'

shutil.copy(source_path, destination_path)
print("Model successfully saved to Google Drive!")
```
